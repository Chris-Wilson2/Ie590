```python
import pandas as pd 
import numpy as np 
from pathlib import Path 

#need to change path to your computers path to the data 
sensors = np.genfromtxt(Path(r"C:\Users\cjwil\Downloads\secom (1)\secom.data"))

labels = np.genfromtxt(Path(r"C:\Users\cjwil\Downloads\secom (1)\secom_labels.data"))
#display(sensors.head())
#display(labels.head())

sensors = pd.DataFrame(sensors)
labels = pd.DataFrame(labels)
print(sensors.shape)
print(labels.shape)

#get only first columns of labels 
labels = labels.iloc[:, 0]

#gets it into combined df 
combine = pd.concat([sensors, labels], axis=1)
print(combine.shape)

#checks -1 (fails) and 1 (passes)
print(combine.iloc[:, -1].value_counts())

#checks NaN values and 0 values in the dataset
print(combine.isna().sum().sum())
print((combine == 0).sum().sum())
```
